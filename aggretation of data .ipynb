```python

master_summary_query = """
SELECT 
    catalog.[Name To Display],
    
    -- May Features (Pull Qty, Calculate Rate, Pull Amount)
    m.Total_Qty AS May_Qty,
    ROUND(m.Amount / m.Total_Qty, 4) AS May_Rate,
    m.Amount AS May_Amount,
    
    -- June Features
    jn.Total_Qty AS June_Qty,
    ROUND(jn.Amount / jn.Total_Qty, 4) AS June_Rate,
    jn.Amount AS June_Amount,
    
    -- July Features
    jl.Total_Qty AS July_Qty,
    ROUND(jl.Amount / jl.Total_Qty, 4) AS July_Rate,
    jl.Amount AS July_Amount

FROM (
    -- Step 1: Collect ALL unique product names across all 3 summaries
    SELECT [Name To Display] FROM raw_may
    UNION
    SELECT [Name To Display] FROM raw_june
    UNION
    SELECT [Name To Display] FROM raw_july
) AS catalog

-- Step 2: Left join the monthly summary tables you uploaded
LEFT JOIN raw_may m ON catalog.[Name To Display] = m.[Name To Display]
LEFT JOIN raw_june jn ON catalog.[Name To Display] = jn.[Name To Display]
LEFT JOIN raw_july jl ON catalog.[Name To Display] = jl.[Name To Display];
"""

```

```python

df_final_matrix = pd.read_sql_query(master_summary_query, conn)


df_final_matrix = df_final_matrix.fillna(0)


df_final_matrix.to_sql("consolidated_market_matrix", conn, index=False, if_exists="replace")


conn.close()

```

```python
NOW WE WILL FIND THE AVG RATE OF ALL THREE COLUMNS
```

```python
import numpy as np

# 1. Isolate the three rate columns and temporarily replace 0 with NaN (Null)
# This prevents 0.0000 from dragging down the average of active months
temp_rates = df_final_matrix[["May_Rate", "June_Rate", "July_Rate"]].replace(0, np.nan)

# 2. Calculate the average across the row (axis=1) ignoring the Null values
# Then fill any completely blank rows back to 0 and round to 4 decimal places
df_final_matrix["Avg_Rate"] = temp_rates.mean(axis=1).fillna(0).round(4)

# Preview the matrix to verify the calculation is exactly as expected
print("🎉 Part 1 Complete: 'Avg_Rate' calculated successfully!")
print(df_final_matrix[["Name To Display", "May_Rate", "June_Rate", "July_Rate", "Avg_Rate"]].head(4))
```

```python
# 1. Drop the individual monthly rate columns as requested
df_final_matrix.drop(columns=["May_Rate", "June_Rate", "July_Rate"], inplace=True)

# 2. Reorder columns to make the dataset clean and logical to read
final_column_order = [
    "Name To Display", 
    "Avg_Rate",
    "May_Qty", "May_Amount", 
    "June_Qty", "June_Amount", 
    "July_Qty", "July_Amount"
]
df_final_matrix = df_final_matrix[final_column_order]

print("🎉 Success: Individual monthly rate columns removed!")
print("👀 Final Clean Matrix Layout:")
print(df_final_matrix.head(5).to_string(index=False))
```

```python
import pandas as pd

# 1. Clean the string names to prevent whitespace mismatching
inv["PRODUCT"] = inv["PRODUCT"].astype(str).str.strip()

# 2. Convert text values into true numbers so Python doesn't throw calculation errors
inv["Current Stock(SellLoose)"] = pd.to_numeric(inv["Current Stock(SellLoose)"], errors='coerce').fillna(0)
inv["PURCHASE VALUE"] = pd.to_numeric(inv["COST RATE"], errors='coerce').fillna(0)
inv["MARGIN%"] = pd.to_numeric(inv["MARGIN%"], errors='coerce').fillna(0)

# 3. Group by the product name and apply different rules for each column
inv_summary = inv.groupby("PRODUCT").agg({
    "Current Stock(SellLoose)": "sum", # Combines physical counts
    "COST RATE": "sum",           # Combines total purchase investment value
    "MARGIN%": "mean"                  # Averages out the percentage profit threshold
}).reset_index()

print("🎉 Part 1 Complete: Warehouse metrics grouped successfully!")
print(inv_summary.head(3))
```

```python
import pandas as pd

# 1. Ensure product string labels are stripped of hidden trailing spaces to guarantee a perfect match
df_final_matrix["Name To Display"] = df_final_matrix["Name To Display"].astype(str).str.strip()
inv_summary["PRODUCT"] = inv_summary["PRODUCT"].astype(str).str.strip()

# 2. Execute the RIGHT JOIN
# This keeps ALL products from df_final_matrix (sales data matrix)
df_turnaround_master = pd.merge(
    inv_summary, 
    df_final_matrix, 
    left_on="PRODUCT", 
    right_on="Name To Display", 
    how="right"
)

# 3. Clean up the join residue
# Drop the duplicate 'PRODUCT' column from inv_summary, since we have 'Name To Display'
df_turnaround_master.drop(columns=["PRODUCT"], inplace=True, errors="ignore")

# 4. Fill physical stock and financial gaps with 0 where there is currently zero warehouse inventory
fill_inventory_cols = ["Current Stock(SellLoose)", "COST RATE", "MARGIN%"]
df_turnaround_master[fill_inventory_cols] = df_turnaround_master[fill_inventory_cols].fillna(0)

# Optional: Clean up data types (Cast stock counts to clean integers)
df_turnaround_master["Current Stock(SellLoose)"] = df_turnaround_master["Current Stock(SellLoose)"].astype(int)

# Round the financial margins and average rates for clean viewing
df_turnaround_master["MARGIN%"] = df_turnaround_master["MARGIN%"].round(2)
df_turnaround_master["Avg_Rate"] = df_turnaround_master["Avg_Rate"].round(2)

print("🎉 Part 1 Complete: Operational datasets unified via a Right Join!")
print(f"• Total Rows Preserved (All Active Selling Products): {len(df_turnaround_master)}")
```

```python
# 1. Map out your exact requested column layout sequence
custom_column_sequence = [
    "Name To Display",            # 1. Product Name first
    "Avg_Rate",                   # 2. Combined Average Rate
    "COST RATE",             # 3. Purchase Value
    "MARGIN%",                    # 4. Profit Margin (%)
    "Current Stock(SellLoose)",   # 5. Current Loose Stock
    "May_Qty", "May_Amount",      # 6. Further monthly columns
    "June_Qty", "June_Amount", 
    "July_Qty", "July_Amount"
]

# 2. Re-index your unified dataframe into this precise column order
df_turnaround_master = df_turnaround_master[custom_column_sequence]

# 3. Clean up formatting and data types for presentation
df_turnaround_master["Current Stock(SellLoose)"] = df_turnaround_master["Current Stock(SellLoose)"].astype(int)
df_turnaround_master["May_Qty"] = df_turnaround_master["May_Qty"].fillna(0).astype(int)
df_turnaround_master["June_Qty"] = df_turnaround_master["June_Qty"].fillna(0).astype(int)
df_turnaround_master["July_Qty"] = df_turnaround_master["July_Qty"].fillna(0).astype(int)

df_turnaround_master["MARGIN%"] = df_turnaround_master["MARGIN%"].round(2)
df_turnaround_master["Avg_Rate"] = df_turnaround_master["Avg_Rate"].round(2)

print("🎉 Part 2 Complete: Master Turnaround Directory structured successfully!")
print("👀 Preview of your customized final layout:")
print(df_turnaround_master.head(5).to_string(index=False))
```

```python
import numpy as np
import pandas as pd

# 1. Clean the 'COST RATE' string parameter into true numeric floats
df_turnaround_master["COST RATE"] = pd.to_numeric(
    df_turnaround_master["COST RATE"], errors="coerce"
).fillna(0.0)

# 2. Round the cost rate parameter to 2 decimal places as requested
df_turnaround_master["COST RATE"] = df_turnaround_master["COST RATE"].round(2)

# 3. Permanently drop the old pre-existing 'MARGIN%' column
if "MARGIN%" in df_turnaround_master.columns:
    df_turnaround_master.drop(columns=["MARGIN%"], inplace=True)

print("🎉 Part 1 Complete: Cost rates cleaned and old margin data purged!")
print(df_turnaround_master[["Name To Display", "Avg_Rate", "COST RATE"]].head(3))
```

```python
#AS THE MARGIN % WAS DIFFERENT WE CALCULATED OUR OWN MARGIN ACCORDING TO SALE AND COST RATE
```

```python
# 4. Mathematically compute Gross Profit Margin % using: ((Price - Cost) / Price) * 100
# np.where handles items with zero sales or zero rates safely to prevent division-by-zero errors
price = df_turnaround_master["Avg_Rate"]
cost = df_turnaround_master["COST RATE"]

calculated_margin = np.where(price > 0, ((price - cost) / price) * 100, 0.0)

# 5. Assign the new values back to a clean 'margin_%' column rounded to 2 decimal places
df_turnaround_master["margin_%"] = np.round(calculated_margin, 2)

# 6. Re-index and sequence columns logically for clean viewing
final_column_layout = [
    "Name To Display",
    "Avg_Rate",
    "COST RATE",
    "margin_%",
    "Current Stock(SellLoose)",
    "May_Qty",
    "May_Amount",
    "June_Qty",
    "June_Amount",
    "July_Qty",
    "July_Amount",
]
df_turnaround_master = df_turnaround_master[final_column_layout]

print("🎉 Part 2 Complete: Custom margin percentages computed safely!")
print("👀 Final Adjusted Table Preview:")
print(df_turnaround_master.head(5).to_string(index=False))
```
